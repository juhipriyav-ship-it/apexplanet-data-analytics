```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Sales"], bins=5)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

plt.figure(figsize=(6,4))
plt.bar(category_sales.index, category_sales.values)
plt.title("Category-wise Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

plt.figure(figsize=(6,6))
plt.pie(region_sales, labels=region_sales.index, autopct="%1.1f%%")
plt.title("Region-wise Sales")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.boxplot(df["Profit"])
plt.title("Profit Distribution")
plt.ylabel("Profit")
plt.show()
```

```python
plt.savefig("../reports/histogram.png")
```

# Observations

- The histogram shows that sales values are distributed across different ranges.
- The bar chart indicates the total sales for each product category, with one category having the highest sales.
- The pie chart shows the percentage contribution of each region to total sales.
- The box plot shows the distribution of profit values and helps identify the spread and any possible outliers.
- Overall, the visualizations provide insights into sales performance, category-wise sales, regional contribution, and profit distribution.
